```python
# Run this cell to view the correlation matrix

# First, prepare numeric columns
df_numeric = df.copy()
df_numeric["quiz_score_avg"] = pd.to_numeric(
    df_numeric["quiz_score_avg"].astype(str).str.replace("%", ""), errors="coerce"
)

numeric_cols = ["age", "hours_spent", "quiz_score_avg", "video_completion_pct",
                "days_since_last_login", "previous_courses_completed", 
                "satisfaction_rating"]
corr_matrix = df_numeric[numeric_cols].corr().round(3)
print(corr_matrix.to_string())
```
